2) Create a dictionary called student , Calculate the average of the marks.Also Display whether the student Passed (average ≥ 50) or Failed.

```python
# Calculate Average and Pass or Fail

student = {
    "name" : "Ali",
    "age" : "16",
    "city" : "Rawalpindi",
    "marks" : [50,60,70]
}

total_marks = 0
tdig = 0

for key, value in student.items():
    print(f"Student {key} is :", value)

for mark in student["marks"]:
    total_marks = total_marks + mark
    tdig = tdig + 1

avg_mark = total_marks/tdig 


#avg_mark = total_marks / len(student["marks"])

print("Total Marks of Student are:", total_marks)
print("Average Marks obtained by Student are:", avg_mark)

# Check whether the student passed or failed
if avg_mark >= 50:
    print("Student Result: Passed")
else:
    print("Student Result: Failed")
```

4) While Loop with User Input, until user to enter numbers -1 . Display number list and  and show highest and lowest number

```python
# List of Number , Find Highest and lowest Number

ulist = []

num= 0

while True:

    num= int(input("Enter any Number: "))

    if num == -1:
        break

    ulist.append(num)


print("Details of Numbers List are :",ulist)

larg_num = ulist[0]
smal_num = ulist[0]

for n in ulist:
    if n>larg_num:
        larg_num = n
    if n<smal_num:
        smal_num = n


print("Largest Number in List are :",larg_num)
print("Smallest Number in List are :",smal_num)
```

5) Word Frequency Counter, Convert the user entered sentence into a list of words.Also Count each word

```python
# Convert the user entered sentence into a list of words


usent = input("Enter a sentence: ")


words = usent.split()


word_count = {} # dictionary create


for w in words:             #word Frequency

    if w in word_count:
        word_count[w] += 1
    else:
        word_count[w] = 1

print("You Entered the Sentence are: \n",usent)
print("Word Frequency are:")

for w, count in word_count.items():
    print(w, ":", count)
```
