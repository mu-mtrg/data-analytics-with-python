4) Count Frequency of Numbers numbers = [1,2,2,3,3,3,4,4,4,4] Create frequency dictionary

```python
# Count dictionary Numbers

numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1

print("Details of Numbers and count of each number are :",frequency)
```

5) Word Counter Take sentence input and count word frequency.

```python
# Word Counter 
word_counter = input("Enter a sentence: ")

words = word_counter.split()

frequency = {}

for w in words:
    if w in frequency:
        frequency[w] = frequency[w] + 1
    else:
        frequency[w] = 1

print("Details of Numbers and count of each number are :", frequency)
```

6) Student Marks & Average marks = {"Math":80, "English":70, "Physics":90} Calculate total and average.

```python
# Calculate total and average MarKs

marks = {
    "Math": 80,
    "English": 70,
    "Physics": 90
}

total = 0
 
for num in marks.values():
    total = total + num

average = total / len(marks)

print("Total Marks are:", total)
print("Average are:", average)
```

7) Highest Scoring Student students = { "Ali":85, "Sara":92, "Ahmed":78, "Zara":95 } Find highest scoring student.

```python
# Highest Scoring Student

students = {
    "Ali" : 85,
    "Sara":92, 
    "Ahmed":78,
    "Zara":95 
    }


std_name = ""
high_marks = 0

for name, num in students.items():
    if num>high_marks:
        high_marks=num
        std_name = name

print("The Highest Scoring Student Name are:", std_name)
print("The Highest Scoring Student Marks are:", high_marks)
```

8) Merge Dictionaries dict1 = {"a":1,"b":2} dict2 = {"c":3,"d":4} Merge and print result.
