1) Write a program that accepts a list from user and prints the alternate elements of the list.

```python
# list Creation and alternate element list

num = []
num_alt = []

user = int(input("How many elements you want to create list ? "))

for i in range(user):
    value = int(input(f"Enter list element {i}: "))
    num.append(value)

print("User Created List:", num)

for i in range(0, len(num), 2):
    #num_alt.append(i)
    num_alt.append(num[i])

print("Alternate List elements are:",num_alt)
    


```

2) Write a program that accepts a list from user and reverses the content of the list. Do not use reverse().

```python
# Reverses the content of the list. Without use reverse().
 
num = []
rnum = []

n = int(input("How many elements you want to create list ? "))

for i in range(n):
    value = int(input(f"Enter list element {i+1}: "))
    num.append(value)

print("User Created List:", num)


for i in range(len(num) - 1, -1,-1): 
    # 1 Arg - len(num) -1, Used fot total list elements
    # 2 Arg - -1 , Used to fetch all list elements, because list start from 0
    # 3 Arg - -1 , Used for reverse counting
    rnum.append(num[i])

print("Reverse List elements are:",rnum)    
```

3) Find Largest Number Without max().

```python
# Largest Number Without max()

num = []


ul = int(input("How many elements you want to create list ? "))

for i in range(ul):
    value = int(input(f"Enter list element {i+1}: "))
    num.append(value)

print("User Created List:", num)

hnum = num[0]

for i in range(1,len(num)):
    if hnum < num[i]:
        hnum = num[i]


print("Largest Number in the List:", hnum)
```

6) Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.  

```python
# Convert the Date from mm/dd/yyyy

date = input("Enter any date in given format (mm/dd/yyyy): ")

month, day, year = date.split("/")

months = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

print("You Entered the Date : ", date)

month_number = int(month)

if 1 <= month_number <= 12:
    month_name = months[month_number - 1] # list index start from Zero
    print(" Your Desire Result of Date : ", month_name, day, "," , year)
else:
    print("Invalid month. Please enter a month between 01 and 12.")



```

7) Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses" 

```python
# create a new string after converting character of each word capitalized

def capitalize_words(sentence):
    words = sentence.split()

    new_sentence = ""

    for word in words:
        new_sentence = new_sentence + word.capitalize() + " "

    return new_sentence


sentence = input("Enter any sentence: ")

result = capitalize_words(sentence)

print(result)
```

8) Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this :

2   11  7   12

5   2   9  15 

8   3   10  42

```python
#sum of each row of matrix

rows = int(input("Enter number of rows in matrix: "))
columns = int(input("Enter number of columns in matrix: "))

matrix = []

for i in range(rows):
    row = []

    for j in range(columns):
        value = int(input(f"Enter value of row {i+1} and Column {j+1} : "))
        row.append(value)

    matrix.append(row)


print("Matrix Values are: ")
for i in range(rows):
    
    for j in range(columns):
        print(matrix[i][j],"\t")

for i in range(rows):
    total = 0

    for j in range(columns):
        total = total + matrix[i][j]

    print(" Matrix Sum of row : ", i + 1, "=", total)
```

10) Write a program to multiply two matrices 

```python
# Program to multiply two matrices 

rows1 = int(input("Enter number of rows in matrix: "))
columns1 = int(input("Enter number of columns in matrix: "))

rows2 = int(input("Enter number of rows of Matrix 2: "))
columns2 = int(input("Enter number of columns of Matrix 2: "))


if columns1 != rows2:

    print("Matrix multiplication is not possible.")

else:

    matrix1 = []
    matrix2 = []
    result = []


    #Matrix 1 values

    for i in range(rows1):
        row = []

        for j in range(columns1):
            value = int(input((f"Enter value of row {x+1} and Column {y+1}  of Matrix 1: ")))
            row.append(value)

        matrix1.append(row)


    #Matrix 2 values

    for x in range(rows2):
        row = []

        for y in range(columns2):
            value = int(input((f"Enter value of row {x+1} and Column {y+1}  of Matrix 2: ")))
            row.append(value)

        matrix2.append(row)


    # Create result matrix

    for i in range(rows1):
        row = []

        for j in range(columns2):

            total = 0

            for k in range(columns1):
                total = total + matrix1[i][k] * matrix2[k][j]

            row.append(total)

        result.append(row)


    print("Result of Multiply of two matrices :")

    for i in range(rows1):
        print(result[i])
```
