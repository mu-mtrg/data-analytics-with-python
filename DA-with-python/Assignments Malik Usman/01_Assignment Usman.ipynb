3. Write a program in which use all the operators we can use in Python

```python
num1 = 17
num2 = 5

# 1. Arithmetic Operators
print("=== Arithmetic Operators ===")
print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
print("Floor Division:", num1 // num2)
print("Modulus:", num1 % num2)
print("Power:", num1 ** num2)


# 2. Assignment Operators
print("\n=== Assignment Operators ===")

num1 += 5
print("Assignment Operators  +=  :", num1)
num1 -= 3
print("Assignment Operators  -= :", num1)

num1*= 2
print("Assignment Operators  *= :", num1)
num1 /= 4
print("Assignment Operators  /= :", num1)

# 3. Comparison Operators
print("\n=== Comparison Operators ===")

print("num1 == num2:", num1 == num2)
print("num1 != num2:", num1 != num2)
print("num1 > num2:", num1 > num2)
print("num1 < num2:", num1 < num2)
print("num1 >= num2:", num1 >= num2)
print("num1 <= num2:", num1 <= num2)

# 4. Logical Operators
print("\n=== Logical Operators ===")

age = 20
print("age > 18 and age < 30:", age > 18 and age < 30)
print("age > 25 or age < 30:", age > 25 or age < 30)
print("not age > 18:", not age > 18)

# 5. Bitwise Operators
print("\n=== Bitwise Operators ===")

print("AND (&):", num1 & num2)
print("OR (|):", num1 | num2)
print("XOR (^):", num1 ^ num2)
print("NOT (~):", ~num1)
print("Left Shift (<<):", num1 << 1)
print("Right Shift (>>):", num1 >> 1)

```

4. Completes the following steps of small task:
    - Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
    - Mention Variable of Total Marks and assign 300 to it
    - Calculate Percentage

```python
english = 85 
islamiat = 90 
maths = 80 

total_marks = 300 

obtained_marks = english + islamiat + maths 

perc = (obtained_marks / total_marks) * 100 

print("English Marks:", english) 
print("Islamiat Marks:", islamiat) 
print("Maths Marks:", maths) 
print("Total Marks:", total_marks) 
print("Obtained Marks:", obtained_marks) 
print("Percentage:", perc, "%")
```

5. Take the values of two variables `a` and `b`, then swap their values without using a third variable and print the result before and after swapping

6. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
radius = 5 
pi = 3.14159

area = pi * radius * radius   # Calculate Area 

circum = 2 * pi * radius  # Calculate Circumference 


# Print the results 
print("Radius of a circle :", radius) 
print("Area of Circle:", area) 
print("Circumference of Circle:", circum)
```

7. Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount

```python
price = 5000
disc_per = 10
disc_amt = (price * disc_per) / 100

final_price = price - disc_per

print("Original Price:", price)
print("Discount Percentage:", disc_per, "%")
print("Discount Amount:", disc_amt)
print("Final Price:", final_price)

```
